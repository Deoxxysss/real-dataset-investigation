### Filtering Columns
#### **What we will filter?**
- **Columns with all zero values**
- **Column with no given values**

```python
#Columns with all Zero values:
# --> There are 18 columns with all rows zero
# From inspection file of ours we get to know those columns that are
All_Zero_Columns = ['M1', 'M2', 'M7', 'M21', 'M22', 'M23', 'F1', 'F2', 'F7', 'F21', 'F22', 'F23', 'JW', 'JB', 'JI', 'JO1', 'JH', 'JN']
```

```python
#Columns with No values:
# --> There is only 1 columns with all NaN Values
# From inspection file of ours we get to know about that columns
Dead_Column = ['CASEID']
```

```python
# Now we create another list of all the columns that we need to remove
All_To_Filter = Dead_Column + All_Zero_Columns
All_To_Filter
```

```python
# Defining the function that will help us drop unwanted column
def drop_dead(df, cols=All_To_Filter):
    return df.drop(columns=[c for c in cols if c in df.columns])
```

```python
# Dropping the columns
df60 = drop_dead(df60)
df61 = drop_dead(df61)
df62 = drop_dead(df62)
```

Cast DIV to str

**DIV** is the only column that needs fixing. It's **category** in all three years,
but the values don't line up — 1960 has "Possessions" and 1961/1962 don't.

When you concat category columns whose value sets disagree, pandas may quietly
downgrade them to plain objects. It depends on the version, which makes it
unpredictable. Casting to **str** up front avoids the whole mess.

I checked **ST** too — its values match across all three years, so no cast
needed there. And **GRP** is already a plain string. Nothing to do.

```python
# Casting 
for df in (df60, df61, df62):
    df['DIV'] = df['DIV'].astype(str)
```

```python
# Changing year from 60--> 1960, 61--> 1961, 62--> 1962
for df in (df60, df61, df62):
    df['YEAR'] = df['YEAR'] + 1900
df60.head()
```

### Now its time to create a new clean dataset combining all three and move toward Report

```python
df = pd.concat([df60, df61, df62], ignore_index=True)
print("Combined:", df.shape)
```

```python
# Lets check other things to be sure
print("Shape:", df.shape)
print("\nYears:")
print(df['YEAR'].value_counts().sort_index())
print("\nNulls:", df.isnull().sum().sum())
print("\nDuplicates on ORI+YEAR+OFF:", df.duplicated(subset=['ORI', 'YEAR', 'OFF']).sum())
print("\nDtypes:")
print(df.dtypes.value_counts())
```

## What we did

Started with three raw **.dta** files straight from ICPSR — 1960, 1961, 1962 — with
identical 72-column schemas but a bunch of problems sitting underneath.

First, we dropped the dead weight. **CASEID** was 100% null, and 18 columns
(**M1**, **M2**, **M7**, **M21**–**M23**, the matching **F** columns, and all six **J**
juvenile-race fields) were zero in every row of every year. That's 19 columns
that carried no information.

We left the reporting flags (**CD1**, **CD2**, **CD3**, **ADJD**) in place. They're
constant in 1960 and 1962, but 1961 has a small number of rows where they
actually vary — about 0.5%. Not enough signal to analyze, but enough that
dropping them would erase a real quirk of the dataset. Kept for the record,
ignored in the analysis.

Then two duplicates in 1960 that shared an **ORI + YEAR + OFF** key. [dropped one of each pair]

**DIV** was **category** in all three years but the values didn't match — 1960
had "Possessions," 1961 and 1962 didn't. Cast it to **str** so pandas wouldn't
silently downgrade the column during concat. **ST** was checked and its
categories matched, so it stayed as-is. **GRP** was already a plain string.

**YEAR** was stored as two-digit (**60**, **61**, **62**). Converted to four digits in
place — **df['YEAR'] + 1900** — so it reads clearly and there's no second year
column cluttering things up.

Finally, stacked the three frames into one and wrote them to **Data/Filtered/**.
Raw files stayed untouched in **Data/Raw/**.

End result: 68,488 rows, 53 columns, one clean CSV ready for analysis.
