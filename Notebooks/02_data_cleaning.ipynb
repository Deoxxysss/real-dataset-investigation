```python
# importing Pandas
import pandas as pd
```

```python
# Reading the .dta files
df60 = pd.read_stata("../Data/Raw/02538-0002-Data1960.dta")
df61 = pd.read_stata("../Data/Raw/02538-0004-Data1961.dta")
df62 = pd.read_stata("../Data/Raw/02538-0006-Data1962.dta")
```

### Filtering Columns
#### **What we will filter?**
- **Columns with all zero values**
- **Column with no given values**

```python
#Columns with all Zero values:
# --> There are 18 columns with all rows zero
# From inspection file of ours we get to know those columns that are
All_Zero_Columns = ['M1', 'M2', 'M7', 'M21', 'M22', 'M23', 'F1', 'F2', 'F7', 'F21', 'F22', 'F23', 'JW', 'JB', 'JI', 'JO1', 'JH', 'JN']
```

```python
#Columns with No values:
# --> There is only 1 columns with all NaN Values
# From inspection file of ours we get to know about that columns
Dead_Column = ['CASEID']
```

```python
# Now we create another list of all the columns that we need to remove
All_To_Filter = Dead_Column + All_Zero_Columns
All_To_Filter
```

```python
# Defining the function that will help us drop unwanted column
def drop_dead(df, cols=All_To_Filter):
    return df.drop(columns=[c for c in cols if c in df.columns])
```

```python
# Dropping the columns
df60 = drop_dead(df60)
df61 = drop_dead(df61)
df62 = drop_dead(df62)
```
