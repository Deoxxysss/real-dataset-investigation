## According to the codebook

### -- Linkage Variables --
- **ST** → Numeric State Code (1–2)
- **ORI** → ORI Code. Originating Agency Identifier (3–9)
- **GRP** → Group. Group 0 is Possessions; 1–7 are cities; 8–9 are counties. Sub-Group (Position 11) is blank when not used. All populations are inclusive. (10–11)
- **DIV** → Division. Geographic division in which the state is located (from 1 thru 9). Possessions are coded "0." (12)

### -- Other Agency Information --
- **YEAR** → Year. Last two digits of the year the data reflects, e.g., "85" = 1985, "90" = 1990, etc. (13–14)
- **MSA** → MSA. Metropolitan Statistical Area (MSA) number in which the city is located, if any. Blank if not used. (15–17)
- **SUB** → Suburban. A "suburban" agency is an MSA city with less than 50,000 population (groups 4–7) together with MBA counties (group 9). 0 = Non-Suburban, 1 = Suburban. (18)
- **JUV** → Report Indication. 0 = Juvenile and Adult Reported, 1 = Juvenile Only Reported, 2 = Adult Only Reported, 3 = Not Reported. (19–19)
- **ADJ** → Adjustment. 1960–79: 0 = Age, Race Reported; 1 = No Race Reported; 2 = No Age Reported. 1980–95: 0 = Age, Race, Ethnic origin Reported; 1 = No Age Reported; 2 = No Race Reported; 3 = No Ethnic origin Reported; 4 = No Race or Ethnic origin Reported; 5 = No Age or Ethnic origin Reported; 6 = No Age or Race Reported. (Other codes should be treated as missing or blank.) (20–20)
- **OFF** → Offense. If value is "000," it is the "header" record currently being defined, otherwise it is the "detail" record. (21–23)
- **SEQ** → Sequence Number. A five-digit number which places all cities in alphabetic order, regardless of state. This field is blank for groups 0, 8, and 9. (24–28)
- **CNTY** → County. Three-digit numeric code for the county in which the agency is located. (29–31)
- **CORE** → Core City. Y = agency is core city of an MSA, N = agency is not a core city of an MSA. (Not fully developed until 1974. Use only for 1974 or later.) (32–32)
- **POP** → Current Population. Total Population for the agency for the year reported. (33–42)
- **AGYCT** → Agency Count. Used to accumulate "agencies used" totals in various tabulations. This field is normally "1" but will be "0" for the U.S. Park Police and all state police agencies whose ORI code ends in "SP" (or "99" in California). (Sometimes A or { in early years, meaning is unclear). (43–43)
- **AGY** → Agency Name. (44–67)
- **STATE** → State Name. (68–73)
- **NMO** → Number of Months Reported. (Missing or blank for 1960–1973. Will vary for 1974–1995 and beyond.) (74–75)

## Missing Values

Null counts per column per year. A "column with nulls" is not automatically a problem — some columns (like `MSA`, `NMO`) are documented as blank under certain conditions. The goal here is to catalogue, not to judge yet.

```python
for name, df in [("1960", df60), ("1961", df61), ("1962", df62)]:
    nulls = df.isnull().sum()
    nulls = nulls[nulls > 0].sort_values(ascending=False)
    print(f"\n=== {name} — columns with nulls ===")
    print(nulls if len(nulls) else "None")
```

```python
# So We now know, while cleaning the data we will have to drop "CASEID" as it is null for all rows
```

```python
# Checking Dublicate Rows:
for name, df in [("1960", df60), ("1961", df61), ("1962", df62)]:
    print(f"{name}: full-row duplicates = {df.duplicated().sum()}")
```

```python
for name, df in [("1960", df60), ("1961", df61), ("1962", df62)]:
    print(f"{name}: OFF-->'000' rows = {(df['OFF'] == '000').sum()}")
```

```python
# Quick pass over each year:
#   - any negative counts (would be impossible in arrest data)
#   - columns that are the same value all the way down (dead weight)
#   - numeric columns that are all zeros (either unused or never reported)

for name, df in [("1960", df60), ("1961", df61), ("1962", df62)]:
    numeric = df.select_dtypes('number')
    print(f"\n----> {name} <-----")

    negs = (numeric < 0).sum().sum()
    print(f"negatives--> {negs}")

    counts = df.nunique()
    dead = counts[counts <= 1].index.tolist()
    print(f"constant cols--> {dead}")

    zeros = [c for c in numeric.columns if (numeric[c] == 0).all()]
    print(f"all-zero cols--> {zeros}")
```
