#### A question can be asked from this
**Which state had the most crime for each year and overall?**

```python
# Lets check the total arrest by gender and different categories
AGE_M = [f'M{i}' for i in range(3, 21) if i != 7]   # M3..M20
AGE_F = [f'F{i}' for i in range(3, 21) if i != 7]   # F3..F20
df['total_male']    = df[AGE_M].sum(axis=1)
df['total_female']  = df[AGE_F].sum(axis=1)
df['total_arrests'] = df['total_male'] + df['total_female']
```

```python
print(f"Total male arrests:   {df['total_male'].sum():,}")
print(f"Total female arrests: {df['total_female'].sum():,}")
print(f"Total arrests:        {df['total_arrests'].sum():,}")
```

```python
# Per-year breakdown
print("\nBy year:")
print(df.groupby('YEAR')[['total_male', 'total_female', 'total_arrests']].sum())
```

```python
# Groupby State
df.groupby('ST')[['total_male', 'total_female', 'total_arrests']].sum()
```

```python
# checking the state with the highest and lowest total arrests in different categories
df.groupby('ST')[['total_male', 'total_female', 'total_arrests']].sum().agg(['idxmax', 'idxmin', 'max', 'min'])
```
