```python
# Age brackets aren't the same width.
#
# M4 is a single year of age (15), but M15 covers five years (25-29).
# A bracket that covers 5 years will naturally collect more arrests than
# one that covers 1 year — not because that age group commits more crime,
# just because there are more birth years inside it.
#
# To compare fairly, divide each bracket's total by how many years it spans.
# That gives arrests per year-of-age, so every bracket is on the same footing.

BRACKET_WIDTH = {
    '3': 2,   # 13-14
    '4': 1,   # 15
    '5': 1,   # 16
    '6': 1,   # 17
    '8': 1,   # 18
    '9': 1,   # 19
    '10': 1,  # 20
    '11': 1,  # 21
    '12': 1,  # 22
    '13': 1,  # 23
    '14': 1,  # 24
    '15': 5,  # 25-29
    '16': 5,  # 30-34
    '17': 5,  # 35-39
    '18': 5,  # 40-44
    '19': 5,  # 45-49
    '20': 5,  # 50-54
    '24': 1,  # no age (adult) — treated as a single bucket
}

age_totals = df[AGE_M].sum()

age_per_year = pd.Series({
    col: age_totals[col] / BRACKET_WIDTH[col[1:]]
    for col in AGE_M
})

age_per_year.sort_values(ascending=False)
```

```python
# Age brackets aren't the same width.
#
# F4 is a single year of age (15), but F15 covers five years (25-29).
# A bracket that covers 5 years will naturally collect more arrests than
# one that covers 1 year — not because that age group commits more crime,
# just because there are more birth years inside it.
#
# To compare fairly, divide each bracket's total by how many years it spans.
# That gives arrests per year-of-age, so every bracket is on the same footing.

BRACKET_WIDTH = {
    '3': 2,   # 13-14
    '4': 1,   # 15
    '5': 1,   # 16
    '6': 1,   # 17
    '8': 1,   # 18
    '9': 1,   # 19
    '10': 1,  # 20
    '11': 1,  # 21
    '12': 1,  # 22
    '13': 1,  # 23
    '14': 1,  # 24
    '15': 5,  # 25-29
    '16': 5,  # 30-34
    '17': 5,  # 35-39
    '18': 5,  # 40-44
    '19': 5,  # 45-49
    '20': 5,  # 50-54
    '24': 1,  # no age (adult) — treated as a single bucket
}

age_totals = df[AGE_F].sum()

age_per_year = pd.Series({
    col: age_totals[col] / BRACKET_WIDTH[col[1:]]
    for col in AGE_F
})

age_per_year.sort_values(ascending=False)
```

## Observation: age bracket ranking looks like a bracket-width artifact

Raw totals rank M20 (50-54) at the top with 1,531,143 arrests, above M4
(15), M5 (16), etc.

But the brackets aren't the same width. M20 covers 5 years of age; M4
covers 1. Without normalizing, we're comparing "arrests per 5-year bracket"
against "arrests per 1-year bracket" — which isn't fair.

**Question for final analysis:** Which age bracket has the highest
arrest rate *per group of age*? Does the ranking change?

Okay so this notebook did its job. We poked around the data, found some weird stuff, and now i have a much better idea of what questions are actually worth asking. Before i start the final analysis notebook, just wanna write down what i found and what the limits are.

### Main things i found

**1. The reporting artifact (this one is important)**
Raw total arrests go up from 1960 to 1962, but thats super misleading. The number of agencies reporting also went up by like 20% in the same time. So we cant just say "crime went up". For the final analysis i need to either look at arrests_per_agency, or only use the agencies that reported in all 3 years. Otherwise the numbers are just lying to us.

**2. Public order stuff is basically half the data**
Drunkenness (230) and disorderly conduct (240) are always at the top. More than half of ALL arrests (55-56%) every single year were public order offenses. And that share barely moved between 60, 61 and 62 which is kinda interesting.

**3. Huge gender gap**
Males get arrested way more than females, like not even close. But the gap changes depending on the offense. For public order and vagrancy stuff its mostly dudes. For larceny (60) the female share is a little higher.

**4. The age bracket thing was a trap**
At first i thought M20 (50-54) had the most arrests. But then i realized M20 covers 5 years of age and M4 only covers 1 year. So of course the bigger bracket has more arrests lol. Once i divided by the number of years in each bracket, the ranking totally changed and younger brackets showed up at the top. Good catch.

### Stuff to watch out for (data limitations)

* The data is arrest-centric not agency-centric. Only 0.19% of rows are zero arrests. So if a row is missing it probably means the agency just didnt report it, not that there were zero arrests. Gotta be careful with agency averages.
* 1961 has some small differences in the reporting flags (CD1-CD3, ADJD) compared to 60 and 62. Not a huge deal but worth remembering.
* No population data in here. So i cant do per-capita rates. Everything is raw counts or per agency.
